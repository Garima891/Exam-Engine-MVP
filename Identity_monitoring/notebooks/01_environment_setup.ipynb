```python
import sys
import cv2
import mediapipe as mp
import numpy as np
import pandas as pd
import sklearn
import matplotlib

print("Python:", sys.version)
print("Python Executable:", sys.executable)

print("\nLibrary Versions:")
print("OpenCV:", cv2.__version__)
print("MediaPipe:", mp.__version__)
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("Scikit-learn:", sklearn.__version__)
print("Matplotlib:", matplotlib.__version__)

print("\nSFace API available:",
      hasattr(cv2, "FaceRecognizerSF"))

print("\nEnvironment verification completed!")
```

```python
from pathlib import Path

# Notebook is inside notebooks/
PROJECT_ROOT = Path.cwd().resolve().parent

landmarker_path = PROJECT_ROOT / "models" / "face_landmarker.task"
sface_path = PROJECT_ROOT / "models" / "face_recognition_sface.onnx"

print("MediaPipe model:", landmarker_path.exists())
print("SFace model:", sface_path.exists())

print("MediaPipe size:", round(landmarker_path.stat().st_size / 1024**2, 2), "MB")
print("SFace size:", round(sface_path.stat().st_size / 1024**2, 2), "MB")
```

Load MediaPipe Face Landmarker

```python
from mediapipe.tasks import python
from mediapipe.tasks.python import vision

base_options = python.BaseOptions(
    model_asset_path=str(landmarker_path)
)

landmarker_options = vision.FaceLandmarkerOptions(
    base_options=base_options,
    running_mode=vision.RunningMode.IMAGE,
    num_faces=2
)

landmarker = vision.FaceLandmarker.create_from_options(
    landmarker_options
)

print("MediaPipe Face Landmarker loaded successfully!")
```

Load OpenCV SFace

```python
recognizer = cv2.FaceRecognizerSF.create(
    str(sface_path),
    ""
)

print("OpenCV SFace loaded successfully!")
```

Verifying model initialization

```python
print("========== MODEL STATUS ==========")

print("MediaPipe:", type(landmarker).__name__)
print("SFace:", type(recognizer).__name__)

print("\nBoth pretrained models initialized successfully!")
```
